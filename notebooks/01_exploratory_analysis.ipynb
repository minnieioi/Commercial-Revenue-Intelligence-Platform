# Commercial Revenue Intelligence — Python Analysis

Goal: turn raw commercial data into account, segment, and revenue insights.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

accounts = pd.read_csv('../data/raw/accounts.csv')
perf = pd.read_csv('../data/raw/monthly_performance.csv', parse_dates=['month'])
opps = pd.read_csv('../data/raw/opportunities.csv', parse_dates=['created_date', 'expected_close_date'])
activities = pd.read_csv('../data/raw/sales_activities.csv', parse_dates=['activity_date'])

accounts.head()
```

## 1. Revenue trend

```python
monthly = perf.groupby('month', as_index=False)['revenue'].sum()
monthly['mom_growth_pct'] = monthly['revenue'].pct_change() * 100
monthly.tail()
```

## 2. Account concentration / Pareto analysis

```python
acct_rev = perf.groupby('account_id', as_index=False)['revenue'].sum().sort_values('revenue', ascending=False)
acct_rev['revenue_share'] = acct_rev['revenue'] / acct_rev['revenue'].sum()
acct_rev['cumulative_share'] = acct_rev['revenue_share'].cumsum()
acct_rev.head(10)
```
